## 8. Loops

```python
primes = [2, 3, 5, 7]
for prime in primes:
    print(prime)

for x in range(5):
    print(x, end=" ")
print()
for x in range(3, 6):
    print(x, end=" ")
print()
for x in range(3, 8, 2):
    print(x, end=" ")
print()
```

```python
count = 0
while count < 5:
    print(count, end=" ")
    count += 1
print()

count = 0
while True:
    print(count, end=" ")
    count += 1
    if count >= 5:
        break
print()

for x in range(10):
    if x % 2 == 0:
        continue
    print(x, end=" ")
print()
```

## 9. Functions

```python
def my_function():
    print("Hello From My Function!")

def my_function_with_args(username, greeting):
    print("Hello, %s, From My Function!, I wish you %s" % (username, greeting))

def sum_two_numbers(a, b):
    return a + b

my_function()
my_function_with_args("John Doe", "a great year!")
x = sum_two_numbers(1, 2)
print(x)
```

**Exercise:** `list_benefits` and `build_sentence`.

```python
def list_benefits():
    return ["More organized code", "More readable code", "Easier code reuse",
            "Allowing programmers to share and connect code together"]

def build_sentence(benefit):
    return "%s is a benefit of functions!" % benefit

def name_the_benefits_of_functions():
    for benefit in list_benefits():
        print(build_sentence(benefit))

name_the_benefits_of_functions()
```

**Exercise:** two vehicles, `car1` (red convertible, $60,000) and `car2` (blue van, $10,000).

```python
class Vehicle:
    name = ""
    kind = "car"
    color = ""
    value = 100.00

    def description(self):
        return "%s is a %s %s worth $%.2f." % (self.name, self.color, self.kind, self.value)

car1 = Vehicle()
car1.name = "Fer"
car1.color = "red"
car1.kind = "convertible"
car1.value = 60000.00

car2 = Vehicle()
car2.name = "Jump"
car2.color = "blue"
car2.kind = "van"
car2.value = 10000.00

print(car1.description())
print(car2.description())
```

**Exercise:** print an alphabetically sorted list of all functions in the `re` module containing the word `find`.

```python
import re

find_members = []
for member in dir(re):
    if "find" in member:
        find_members.append(member)

print(sorted(find_members))
```
