## 1. Generators

```python
import random

def lottery():
    # returns 6 numbers between 1 and 40
    for i in range(6):
        yield random.randint(1, 40)
    # returns a 7th number between 1 and 15
    yield random.randint(1, 15)

random.seed(5831)
for random_number in lottery():
    print("And the next number is... %d!" % random_number)
```

**Exercise:** a generator that returns the Fibonacci series.

```python
def fib():
    a, b = 1, 1
    while True:
        yield a
        a, b = b, a + b

import types
if isinstance(fib(), types.GeneratorType):
    print("Good, The fib function is a generator.")

counter = 0
for n in fib():
    print(n)
    counter += 1
    if counter == 10:
        break
```

## 2. List Comprehensions

```python
sentence = "the quick brown fox jumps over the lazy dog"
words = sentence.split()

word_lengths = []
for word in words:
    if word != "the":
        word_lengths.append(len(word))
print(word_lengths)

word_lengths = [len(word) for word in words if word != "the"]
print(word_lengths)
```

## 4. Multiple Function Arguments

```python
def foo(first, second, third, *therest):
    print("First: %s" % first)
    print("Second: %s" % second)
    print("Third: %s" % third)
    print("And all the rest... %s" % list(therest))

foo(1, 2, 3, 4, 5)
```

```python
def bar(first, second, third, **options):
    if options.get("action") == "sum":
        print("The sum is: %d" % (first + second + third))
    if options.get("number") == "first":
        return first

result = bar(1, 2, 3, action="sum", number="first")
print("Result: %d" % result)
```

**Exercise:** `foo` returns the number of extra arguments; `bar` returns True if `magicnumber` is 7.

```python
def foo(a, b, c, *args):
    return len(args)

def bar(a, b, c, **kwargs):
    return kwargs.get("magicnumber") == 7

if foo(1, 2, 3, 4) == 1:
    print("Good.")
if foo(1, 2, 3, 4, 5) == 2:
    print("Better.")
if bar(1, 2, 3, magicnumber=6) == False:
    print("Great.")
if bar(1, 2, 3, magicnumber=7) == True:
    print("Awesome!")
```

## 6. Exception Handling

```python
def do_stuff_with_number(n):
    print(n)

def catch_this():
    the_list = (1, 2, 3, 4, 5)
    for i in range(20):
        try:
            do_stuff_with_number(the_list[i])
        except IndexError:
            do_stuff_with_number(0)

catch_this()
```

**Exercise:** handle the exception so the last name is returned as "Smith" (with `get_last_name`).

```python
actor = {"name": "John Cleese", "rank": "awesome"}

def get_last_name():
    return actor["name"].split()[1]

try:
    print(actor["last_name"])
except KeyError:
    print("KeyError handled - falling back to splitting the name")
    print("All exceptions caught! Good job!")
    print("The actor's last name is %s" % get_last_name())
```

**Exercise:** add the key-value pair `"Me": 800` to the JSON string and print it.

```python
def add_employee(salaries_json, name, salary):
    salaries = json.loads(salaries_json)
    salaries[name] = salary
    return json.dumps(salaries)

salaries = '{"Alfred" : 300, "Jane" : 400 }'
new_salaries = add_employee(salaries, "Me", 800)
decoded_salaries = json.loads(new_salaries)
print(decoded_salaries["Alfred"])
print(decoded_salaries["Jane"])
print(decoded_salaries["Me"])
```

## 11. Closures

```python
def transmit_to_space(message):
    def data_transmitter():
        print(message)
    data_transmitter()

transmit_to_space("Test message")
```

```python
def print_msg(number):
    def printer():
        nonlocal number
        number = 3
        print(number)
    printer()
    print(number)

print_msg(9)
```

```python
def transmit_to_space(message):
    def data_transmitter():
        print(message)
    return data_transmitter

fun2 = transmit_to_space("Burn the Sun!")
fun2()
```

## 12. Decorators

```python
def double_out(old_function):
    def new_function(*args, **kwds):
        return 2 * old_function(*args, **kwds)
    return new_function

def double_ii(old_function):
    def new_function(arg):
        return old_function(arg * 2)
    return new_function

@double_out
def add_one(x):
    return x + 1

@double_ii
def square(x):
    return x * x

print(add_one(3))   # 2 * (3 + 1)
print(square(3))    # (3 * 2) ** 2
```

```python
def check(old_function):
    def new_function(arg):
        if arg < 0:
            raise ValueError("Negative Argument")
        return old_function(arg)
    return new_function

@check
def root(x):
    return x ** 0.5

print(root(16))
try:
    root(-1)
except ValueError as e:
    print("ValueError:", e)
```

**Exercise:** a decorator factory `type_check(correct_type)` that only runs the function if the input has the right type.

```python
def type_check(correct_type):
    def check(old_function):
        def new_function(arg):
            if isinstance(arg, correct_type):
                return old_function(arg)
            print("Bad Type")
        return new_function
    return check

@type_check(int)
def times2(num):
    return num * 2

print(times2(2))
times2("Not A Number")

@type_check(str)
def first_letter(word):
    return word[0]

print(first_letter("Hello World"))
first_letter(["Not", "A", "String"])
```

## 13. Map, Filter, Reduce

```python
my_pets = ["alfred", "tabitha", "william", "arla"]
print(list(map(str.upper, my_pets)))

circle_areas = [3.56773, 5.57668, 4.00914, 56.24241, 9.01344, 32.00013]
print(list(map(round, circle_areas, range(1, 7))))

print(list(zip(["a", "b", "c"], [1, 2, 3])))
```

```python
scores = [66, 90, 68, 59, 76, 60, 88, 74, 81, 65]
print(list(filter(lambda s: s > 75, scores)))

dromes = ("demigod", "rewire", "madam", "freer", "anutforajaroftuna", "kiosk")
print(list(filter(lambda w: w == w[::-1], dromes)))
```

```python
from functools import reduce

numbers = [3, 4, 6, 9, 34, 12]
print(reduce(lambda a, b: a + b, numbers))
print(reduce(lambda a, b: a + b, numbers, 10))
```

**Exercise:** use map, filter and reduce together.

```python
from functools import reduce

my_floats = [4.35, 6.09, 3.25, 9.77, 2.16, 8.88, 4.59]
my_names = ["olumide", "akinremi", "josiah", "temidayo", "omoseun"]
my_numbers = [4, 6, 9, 23, 5]

map_result = list(map(lambda x: round(x ** 2, 3), my_floats))
filter_result = list(filter(lambda name: len(name) <= 7, my_names))
reduce_result = reduce(lambda num1, num2: num1 * num2, my_numbers, 1)

print(map_result)
print(filter_result)
print(reduce_result)
```
