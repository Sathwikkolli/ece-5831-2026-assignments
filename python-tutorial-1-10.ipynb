## Section 5. Functions
### Defining and calling functions

```python
def greet(name):
    return f"Hi {name}"

greeting = greet("John")
print(greeting)

def sum_(start, end):
    total = 0
    for n in range(start, end + 1):
        total += n
    return total

print(sum_(1, 10))
```

### Default parameters

```python
def greet(name, message="Hi"):
    return f"{message} {name}"

print(greet("John"))
print(greet("Jane", "Hello"))
```

### Keyword arguments

```python
def get_net_price(price, tax=0.07, discount=0.05):
    return price * (1 + tax - discount)

print(round(get_net_price(100), 2))
print(round(get_net_price(100, discount=0.06), 2))
print(round(get_net_price(price=100, tax=0.08, discount=0.06), 2))
```

### Recursive functions

```python
def count_down(start):
    """ Count down from a number """
    print(start, end=" ")
    next_ = start - 1
    if next_ > 0:
        count_down(next_)

count_down(3)
print()

def sum_(n):
    if n > 0:
        return n + sum_(n - 1)
    return 0

print(sum_(100))
```

### Lambda expressions

```python
def get_full_name(first_name, last_name, formatter):
    return formatter(first_name, last_name)

print(get_full_name("John", "Doe", lambda first, last: f"{first} {last}"))
print(get_full_name("John", "Doe", lambda first, last: f"{last} {first}"))

def times(n):
    return lambda x: x * n

double = times(2)
print(double(2), double(3))

callables = [lambda x, i=i: i * x for i in range(1, 4)]
for f in callables:
    print(f(2), end=" ")
```

### reduce()

```python
from functools import reduce

scores = [75, 65, 80, 95, 50]
print(reduce(lambda a, b: a + b, scores))

# how it works, step by step
def sum_(a, b):
    print(f"a={a}, b={b}, {a} + {b} = {a + b}")
    return a + b

print(reduce(sum_, scores))
```

### Union, intersection, difference, symmetric difference

```python
s1 = {"Python", "Java"}
s2 = {"C#", "Java"}
print("union:", s1 | s2, s1.union(s2))
print("intersection:", s1 & s2, s1.intersection(s2))
print("difference:", s1 - s2, s1.difference(s2))
print("symmetric difference:", s1 ^ s2, s1.symmetric_difference(s2))
```

### Subset, superset, disjoint

```python
numbers = {1, 2, 3, 4, 5}
scores = {1, 2, 3}
print(scores.issubset(numbers), scores <= numbers, scores < numbers)
print(numbers.issuperset(scores), numbers >= scores)

odd = {1, 3, 5}
even = {2, 4, 6}
print(odd.isdisjoint(even))
```

## Section 9. Exception handling
### try...except

```python
def divide(a, b):
    try:
        result = a / b
        print(f"{a} / {b} = {result}")
    except ZeroDivisionError:
        print("The current sales cannot be 0")
    except (ValueError, TypeError) as error:
        print("Error:", error)

divide(10, 2)
divide(10, 0)
divide(10, "x")
```

### try...except...finally

```python
a, b = 10, 0
try:
    c = a / b
    print(c)
except ZeroDivisionError as error:
    print(error)
finally:
    print("Finishing up.")
```

### try...except...else

```python
def calculate_bmi(height, weight):
    """ calculate body mass index (BMI) """
    return weight / height ** 2

def evaluate_bmi(bmi):
    if 18.5 <= bmi <= 24.9:
        return "healthy"
    elif bmi >= 25:
        return "overweight"
    return "underweight"

for height, weight in [("1.75", "70"), ("abc", "70")]:
    try:
        h = float(height)
        w = int(weight)
    except ValueError as error:
        print("Error, please enter a valid number:", error)
    else:
        bmi = calculate_bmi(h, w)
        print(f"BMI {bmi:.1f} -> {evaluate_bmi(bmi)}")
```

## Section 10. More on Python loops
### for...else

```python
people = [{"name": "John", "sex": "Male"}, {"name": "Jane", "sex": "Female"}]

for name in ["Jane", "Peter"]:
    for person in people:
        if person["name"] == name:
            print(f"{name} is {person['sex']}")
            break
    else:
        print(f"{name} not found!")
```

### while...else

```python
basket = [{"fruit": "apple", "qty": 20}, {"fruit": "banana", "qty": 30}]

for fruit in ["banana", "orange"]:
    index = 0
    while index < len(basket):
        if basket[index]["fruit"] == fruit:
            print(f"{fruit}: {basket[index]['qty']}")
            break
        index += 1
    else:
        basket.append({"fruit": fruit, "qty": 10})
        print(f"{fruit} not found - added with qty 10")
print(basket)
```

### Emulating do...while

```python
import random

random.seed(42)
secret_number = random.randint(0, 10)
attempt = 0
# "guesses" come from a list instead of input() so the notebook runs non-interactively
guesses = iter(range(11))
while True:
    attempt += 1
    guess = next(guesses)
    if guess > secret_number:
        print(f"{guess}: too high")
    elif guess < secret_number:
        print(f"{guess}: too low")
    else:
        print(f"Bingo! {guess} found in {attempt} attempts.")
        break
```
